```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score, precision_score,
    recall_score, f1_score, confusion_matrix
)

# Training data: [Study Hours, Attendance Percentage]
X = np.array([
    [2, 50],
    [3, 60],
    [4, 65],
    [5, 70],
    [6, 75],
    [7, 80],
    [8, 85],
    [9, 90]
])

# 0 = Fail, 1 = Pass
y = np.array([0, 0, 0, 1, 1, 1, 1, 1])

# Train Decision Tree
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# Test data
X_test = np.array([
    [3, 55],
    [5, 72],
    [8, 88],
    [4, 60]
])

y_test = np.array([0, 1, 1, 0])

# Prediction
y_pred = model.predict(X_test)

# Evaluation
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("----- Decision Tree Classification -----")
print("Actual Labels:", y_test)
print("Predicted Labels:", y_pred)
print("Confusion Matrix:\n", cm)
print("Accuracy:", accuracy)
print("Error Rate:", error_rate)
print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)

# Decision Tree Visualization
plt.figure(figsize=(10, 6))
plot_tree(
    model,
    feature_names=["Study Hours", "Attendance"],
    class_names=["Fail", "Pass"],
    filled=True,
    rounded=True
)
plt.title("Decision Tree - Student Pass/Fail")
plt.show()

# Confusion Matrix
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Fail", "Pass"],
            yticklabels=["Fail", "Pass"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Decision Tree Confusion Matrix")
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score, precision_score,
    recall_score, f1_score, confusion_matrix
)

# Training data
X_train = np.array([
    [2, 50],
    [3, 60],
    [4, 65],
    [5, 70],
    [6, 75],
    [7, 80],
    [8, 85],
    [9, 90]
])

y_train = np.array([0, 0, 0, 1, 1, 1, 1, 1])

# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

# Test data
X_test = np.array([
    [3, 55],
    [5, 72],
    [8, 88],
    [4, 60]
])

y_test = np.array([0, 1, 1, 0])

# Prediction
y_pred = model.predict(X_test)

# Performance evaluation
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("----- KNN Classification -----")
print("Actual Labels:", y_test)
print("Predicted Labels:", y_pred)
print("Confusion Matrix:\n", cm)
print("Accuracy:", accuracy)
print("Error Rate:", error_rate)
print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)

# Confusion Matrix
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Greens",
            xticklabels=["Fail", "Pass"],
            yticklabels=["Fail", "Pass"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("KNN Confusion Matrix")
plt.show()

# Training Data Visualization
plt.figure(figsize=(8, 5))
plt.scatter(
    X_train[:, 0], X_train[:, 1],
    c=y_train, cmap="coolwarm",
    s=100, edgecolors="black"
)

plt.xlabel("Study Hours")
plt.ylabel("Attendance Percentage")
plt.title("KNN - Student Pass/Fail Classification")
plt.grid(True)
plt.show()
```
