```python

import numpy as np
from scipy.stats import bernoulli, binom
import matplotlib.pyplot as plt

# Bernoulli Distribution
print("----- Bernoulli Distribution -----")

p = 0.7
rng = np.random.default_rng(42)

samples = bernoulli.rvs(p, size=10, random_state=rng)

print("Probability of Passing:", p)
print("Student Outcomes (1=Pass, 0=Fail):", samples)
print("Probability of Pass:", bernoulli.pmf(1, p))
print("Probability of Fail:", round(bernoulli.pmf(0, p), 2))

# Binomial Distribution
print("\n----- Binomial Distribution -----")

n = 10
k = 8

probability = binom.pmf(k, n, p)

print("Number of Students:", n)
print("Probability of Passing:", p)
print("Probability exactly 8 pass:", probability)

outcomes = binom.rvs(n, p, size=5, random_state=rng)
print("Random Binomial Outcomes:", outcomes)

# Visualization
x = np.arange(n + 1)
y = binom.pmf(x, n, p)

plt.figure(figsize=(8, 4))
plt.bar(x, y, color="skyblue", edgecolor="black")
plt.title("Binomial Distribution - Student Pass Analysis")
plt.xlabel("Number of Students Passed")
plt.ylabel("Probability")
plt.xticks(x)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python

import numpy as np
from scipy import stats
from statsmodels.stats.weightstats import ztest
import matplotlib.pyplot as plt

alpha = 0.05

# Z-Test
print("----- Z-Test -----")

sample_marks = np.array([78, 82, 85, 88, 90, 76, 95, 89, 84, 91])
population_mean = 80

z_stat, p_value = ztest(sample_marks, value=population_mean)

print("Sample Marks:", sample_marks)
print("Sample Mean:", np.mean(sample_marks))
print("Population Mean:", population_mean)
print("Z-Statistic:", round(z_stat, 4))
print("P-Value:", round(p_value, 4))

if p_value < alpha:
    print("Result: Reject Null Hypothesis")
else:
    print("Result: Fail to Reject Null Hypothesis")

# Independent t-Test
print("\n----- Independent t-Test -----")

class_A = np.array([78, 82, 85, 88, 90])
class_B = np.array([72, 75, 70, 68, 74])

t_stat, p_value = stats.ttest_ind(class_A, class_B)

print("Class A Marks:", class_A)
print("Class B Marks:", class_B)
print("Class A Mean:", np.mean(class_A))
print("Class B Mean:", np.mean(class_B))
print("t-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))

if p_value < alpha:
    print("Result: Reject Null Hypothesis")
else:
    print("Result: Fail to Reject Null Hypothesis")

# Visualization
plt.figure(figsize=(7, 4))
plt.bar(["Class A", "Class B"],
        [np.mean(class_A), np.mean(class_B)],
        color=["royalblue", "orange"])

plt.title("Comparison of Class A and Class B")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.grid(axis="y", alpha=0.3)
plt.show()
```
