```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

print("----- Linear Regression -----")

# Training dataset
X_linear = np.array([[1000], [1200], [1500], [1800], [2000]])
y_linear = np.array([50, 55, 65, 80, 90])

# Create and train model
linear_model = LinearRegression()
linear_model.fit(X_linear, y_linear)

# Predict price for a 1700 sq.ft house
new_house_size = np.array([[1700]])
predicted_price = linear_model.predict(new_house_size)

print("Training House Sizes:", X_linear.flatten())
print("Training House Prices:", y_linear)
print("Predicted Price for 1700 sq.ft house:",
      round(predicted_price[0], 2), "Lakhs")

# Plot graph
plt.figure(figsize=(8, 5))
plt.scatter(X_linear, y_linear, color="blue", label="Actual Prices")
plt.plot(X_linear, linear_model.predict(X_linear),
         color="red", label="Regression Line")
plt.scatter(1700, predicted_price[0],
            color="green", s=120, label="Predicted Price")

plt.xlabel("House Size (sq.ft)")
plt.ylabel("House Price (Lakhs)")
plt.title("Linear Regression - House Price Prediction")
plt.legend()
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

print("----- Logistic Regression -----")

# Training dataset
X_logistic = np.array([[40], [50], [60], [70], [80], [90]])
y_logistic = np.array([0, 0, 0, 1, 1, 1])

# Create and train model
logistic_model = LogisticRegression()
logistic_model.fit(X_logistic, y_logistic)

# Predict purchase for 75 lakh house
test_price = np.array([[75]])

purchase_prediction = logistic_model.predict(test_price)
purchase_probability = logistic_model.predict_proba(test_price)

print("Training House Prices:", X_logistic.flatten())
print("Purchase Labels:", y_logistic)
print("Test House Price:", test_price[0][0])

print("Purchase Prediction:", purchase_prediction[0])
print("Prediction Probabilities:", purchase_probability)

if purchase_prediction[0] == 1:
    print("Result: Customer will purchase the house")
else:
    print("Result: Customer will not purchase the house")

# Plot graph
x_values = np.linspace(35, 95, 200).reshape(-1, 1)
probabilities = logistic_model.predict_proba(x_values)[:, 1]

plt.figure(figsize=(8, 5))
plt.scatter(X_logistic, y_logistic,
            color="red", label="Actual Data")
plt.plot(x_values, probabilities,
         color="blue", label="Logistic Curve")

plt.axhline(y=0.5, color="gray", linestyle="--")
plt.scatter(75, purchase_probability[0][1],
            color="green", s=120, label="Prediction")

plt.xlabel("House Price (Lakhs)")
plt.ylabel("Purchase Probability")
plt.title("Logistic Regression - House Purchase Prediction")
plt.legend()
plt.grid(True)
plt.show()
```
